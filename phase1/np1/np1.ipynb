```python
df["total_sms"] = df["sms_in"] + df["sms_out"]

df["total_calls"] = df["call_in"] + df["call_out"]

df["total_activity"] = (
    df["total_sms"]
    + df["total_calls"]
    + df["internet_activity"]
)
```

```python
hourly_activity = (
    df.groupby("timestamp")[
        activity_cols
    ]
    .sum()
)
```

```python
hourly_activity["total_activity"] = hourly_activity.sum(axis=1)
```

```python
busiest_hour = hourly_activity["total_activity"].idxmax()

print(busiest_hour)
print(hourly_activity.loc[busiest_hour])
```

```python
grid_activity = (
    df.groupby("grid_id")[
        activity_cols
    ]
    .sum()
)

grid_activity["total_activity"] = grid_activity.sum(axis=1)
```

```python
busiest_grid = grid_activity["total_activity"].idxmax()

print(busiest_grid)
print(grid_activity.loc[busiest_grid])
```

### Network Analytics Team — Data Profiling Summary

* The supplied file contains **1,891,928 raw activity records** covering **10,000 geographic grid cells** and **24 hourly timestamps** from **2013-11-01 00:00 to 23:00**. The timestamps have a verified **one-hour cadence**.

* The raw data grain is **timestamp + grid_id + country_code** rather than simply timestamp + grid_id. For example, grid 1 at 00:00 appears in three records with country codes **0, 33, and 39**. The country-code dimension must therefore be aggregated before producing the hourly grid analytics layer.

* The activity fields contain substantial blank values: `sms_in` 1,086,153; `sms_out` 1,422,446; `call_in` 1,407,781; `call_out` 1,037,413; and `internet_activity` 1,087,074. These blanks must be handled through the documented curated-layer null policy rather than silently removing otherwise valid records.

* Data-quality checks found **no missing grid IDs or timestamps, no exact duplicate records, and no negative activity values**. All observed grid IDs fall within the expected **1–10,000** range. The dataset contains **246 country-code categories**.

* After aggregating activity across country-code rows, the highest overall hourly activity occurred at **11:00**, with approximately **5.19 million total activity units**, while **grid 5161** had the highest cumulative activity at approximately **274,800 activity units** for the supplied day. These values represent communication activity measures and should not be interpreted as literal message counts, call counts, megabytes, or network congestion.
